```python
import pandas as pd
import glob
import os

pasta_csv = r"C:\Users\marcu\OneDrive\Área de Trabalho\CSV"

arquivos = glob.glob(os.path.join(pasta_csv, "*.csv"))
arquivos += glob.glob(os.path.join(pasta_csv, "*.CSV"))

print(f"Arquivos encontrados: {len(arquivos)}")

for arquivo in arquivos:

    print(f"Convertendo: {os.path.basename(arquivo)}")

    df = pd.read_csv(
        arquivo,
        sep=";",
        decimal=","
    )

    df.columns = ["Numero_Onda", "Absorbancia"]

    nome_saida = os.path.join(
        pasta_csv,
        os.path.splitext(os.path.basename(arquivo))[0]
        + "_tratado.csv"
    )

    df.to_csv(
        nome_saida,
        index=False
    )

print("Conversão concluída.")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import glob
import os

# Pasta contendo os arquivos tratados
pasta_csv = r"C:\Users\marcu\OneDrive\Área de Trabalho\CSV"

# Busca apenas os arquivos tratados
arquivos = sorted(
    glob.glob(os.path.join(pasta_csv, "*_tratado.csv"))
)

# Seleciona apenas os 10 primeiros
arquivos = arquivos[:10]

plt.figure(figsize=(12, 8))

for arquivo in arquivos:

    df = pd.read_csv(arquivo)

    plt.plot(
        df["Numero_Onda"],
        df["Absorbancia"],
        linewidth=1.5,
        label=os.path.basename(arquivo).replace("_tratado.csv", "")
    )

# Padrão FTIR
plt.gca().invert_xaxis()

plt.xlabel("Número de onda (cm⁻¹)", fontsize=12)
plt.ylabel("Absorbância", fontsize=12)
plt.title("Espectros FTIR/MIR - Primeiras 10 Amostras", fontsize=14)

plt.grid(True, alpha=0.3)

# Legenda organizada
plt.legend(
    title="Amostras",
    fontsize=9,
    title_fontsize=10,
    loc="best"
)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import glob
import os

pasta_csv = r"C:\Users\marcu\OneDrive\Área de Trabalho\CSV"

arquivos = sorted(
    glob.glob(os.path.join(pasta_csv, "*_tratado.csv"))
)

print(f"Total de amostras: {len(arquivos)}")

plt.figure(figsize=(14,8))

for arquivo in arquivos:

    try:

        df = pd.read_csv(arquivo)

        # Usa as duas primeiras colunas do arquivo
        x = df.iloc[:,0]
        y = df.iloc[:,1]

        plt.plot(
            x,
            y,
            linewidth=0.5,
            alpha=0.4
        )

    except Exception as erro:
        print(f"Erro em {os.path.basename(arquivo)}")
        print(erro)

plt.gca().invert_xaxis()

plt.xlabel("Número de onda (cm⁻¹)")
plt.ylabel("Absorbância")
plt.title("Espectros FTIR/MIR de Todas as Amostras")

plt.grid(True, alpha=0.3)

plt.tight_layout()

plt.show()
```
